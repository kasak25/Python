### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def Prime(n):
    for i in range(2,(n//2)+1):
        if n%i == 0:
            return 0

    return 1
Prime(9)
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def prime_range(n,m):
    l = []
    for i in range(n,m+1):
       if Prime(i) :
           l.append(i)
    print(l)
prime_range(5,40)
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def palindrome(s):
    if s[::1] == s[::-1]:
        print('is palindrome')
    else:
        print('not palindrome')
        
palindrome('kasak')
```

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def sumOft(li):
    sum = 0
    for i in li:
        sum = sum + i[0]
    return sum
        
sumOft([(1,2),(5,6),(4,2)])
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

n = int(input("Enter n: "))
print("Nth Fibonacci term:", fibonacci(n))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
def scores(l):
    sum = 0 
    for i in l:
        if i%10 == 0:
            sum = sum + i
    return sum
scores([200, 456, 300, 100, 234, 678])
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invert_dict(d):
    return {v: k for k, v in d.items()}

dict1 = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
d = invert_dict(dict1)
print("Before:", dict1)
print("After:", d)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_case(s):
    no_upper = sum(1 for c in s if c.isupper())
    no_lower = sum(1 for c in s if c.islower())
    return no_upper, no_lower

s1 = "AbcDEfgh"
upper, lower = count_case(s1)
print("no_upper =", upper, ", no_lower =", lower)
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
def sum(a,b):
    return a+b
print('positional Arguments:-',sum(4,2))
print('Keyword Arguments:-',sum(b=2,a=2))

def default(a,b=0):
    return a+b
print('Default Arguments:-',default(5))

def sum2(a,*b):
    s = a
    for i in b:
        s += i
    return s
print('Variable Length Positional(*args) Arguments:-',sum2(2,5,6))

def sum3(a,**b):
    s2 = a
    for i,j in b.items():
        s2 += j
    return s2
print('Variable Length keyword(*args) Arguments:-',sum3(9,b=4))
def sum4(a,b,/):
    return a+b
print('keyword-only Arguments:-',sum4(4,8))
def sum5(*,a,b):
    return a+b
print('positional-only Arguments:-',sum5(a=7,b=2))
```

### 10) WAP to test if the given tuple is distinct or not.
**Sample Input:**
```
tuple1 = (1, 2, 3, 4)
tuple2 = (1, 2, 2, 3)
```
**Sample Output:**
```
(1, 2, 3, 4) is distinct? -> True
(1, 2, 2, 3) is distinct? -> False
```

````python
tuple1 = (1, 2, 3, 4)
new_t1 = ()
tuple2 = (1, 2, 2, 3)
new_t2 = ()
for i in tuple1:
    if i not in new_t1:
        new_t1 += (i,)
for i in tuple2:
    if i not in new_t2:
        new_t2 += (i,)
if(new_t1 == tuple1):
    print('tuple 1 is distinct -> true')
else:
    print('tuple 1 is not distinct -> false')
if(new_t2 == tuple2):
    print('tuple 2 is distinct -> true')
else:
    print('tuple 2 is not distinct -> false')
````

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])

result = [sum(i) for i in zip(*T)]
print(*result)
````

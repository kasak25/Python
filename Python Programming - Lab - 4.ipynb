### 16) WAP to print all the perfect numbers between 1 to n.

```python
n = int(input("enter number:"))
for i in range(1,n):
    sum = 0
    for j in range(1,i):
        if(i%j == 0):
            sum = sum + j
    if(sum == i):
        print(i)
```
